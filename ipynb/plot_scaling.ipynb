# Scaling Plot Setup

This notebook builds a simple scaling relation using float-native arrays.


```python
import sys
from pathlib import Path

def _find_repo_root():
    for start in (Path.cwd(), Path.cwd().resolve()):
        for candidate in (start, *start.parents):
            if (candidate / 'pysrc' / 'solve_ode.py').exists():
                return candidate
    raise RuntimeError('Could not find repo root containing pysrc/solve_ode.py')

repo_root = _find_repo_root()
pysrc_path = repo_root / 'pysrc'
if str(pysrc_path) not in sys.path:
    sys.path.insert(0, str(pysrc_path))

import numpy as np

import matplotlib.pyplot as plt
```

```python
vc_kms = np.linspace(80.0, 220.0, 32)
mdot_hot_max = 0.7 * (vc_kms / 100.0) ** 5.4
fig, ax = plt.subplots()
ax.plot(vc_kms, mdot_hot_max)
ax.set_xlabel('v_c [km/s]')
ax.set_ylabel('Mdot_hot,max [Msun/yr]')
scaled_values = {'vc_min': float(vc_kms.min()), 'vc_max': float(vc_kms.max())}
plt.close(fig)
scaled_values
```
