```python
import matplotlib.pyplot as plt
import numpy as np

# Scaling test data
nodes = np.array([1, 2, 4, 8, 16, 32])
performance = np.array([1.237794e+08, 2.321966e+08, 4.642498e+08, 
                       7.990371e+08, 1.278067e+09, 2.277784e+09])

# Create the log-log plot
plt.figure(figsize=(10, 7))

# Plot the actual data
plt.loglog(nodes, performance, 'o-', linewidth=2.5, markersize=8, 
           color='#2563eb', label='Actual Performance', markerfacecolor='white',
           markeredgecolor='#2563eb', markeredgewidth=2)

# Create optimal scaling reference line (perfect linear scaling)
# Use the first data point as reference and scale linearly
optimal_performance = performance[0] * nodes
plt.loglog(nodes, optimal_performance, '--', linewidth=1.5, 
           color='lightgray', alpha=0.7, label='Optimal Scaling')

# Customize the plot
plt.xlabel('Number of Nodes', fontsize=12)
plt.ylabel('Zone-cycles/cpu_second', fontsize=12)
plt.title('Scaling Test: Performance vs Number of Nodes', fontsize=14, pad=20)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)

# Set axis limits and ticks
plt.xlim(0.8, 40)
plt.ylim(1e8, 5e9)

# Add minor grid
plt.grid(True, which="minor", alpha=0.2)
plt.tight_layout()
plt.show()

# Print scaling efficiency for each configuration
print("Scaling Analysis:")
print("-" * 40)
print(f"{'Nodes':>5} {'Performance':>12} {'Efficiency':>10}")
print("-" * 40)
for i, (n, p) in enumerate(zip(nodes, performance)):
    if i == 0:
        efficiency = 100.0
    else:
        efficiency = (p / performance[0]) / n * 100
    print(f"{n:>5} {p:>12.2e} {efficiency:>9.1f}%")
```
